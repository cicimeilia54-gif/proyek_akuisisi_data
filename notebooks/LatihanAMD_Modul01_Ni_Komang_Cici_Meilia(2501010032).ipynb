## Langkah 1. Verifikasi versi Python dan library inti

```python
import sys, platform
print("Python :", sys.version)
print("Platform:", platform.platform())
for lib in ["pandas", "numpy", "requests", "bs4"]:
    try:
        m = __import__(lib)
        print(f"{lib:10s}: {getattr(m, '__version__', 'terpasang')}")
    except ImportError:
        print(f"{lib:10s}: BELUM TERPASANG -> pip install {lib}")
```

## Langkah 2. Membangun struktur folder proyek

```python
from pathlib import Path
ROOT = Path("proyek_akuisisi_data")
FOLDERS = [
"data/raw", "data/interim", "data/processed",
"notebooks", "src", "docs", "reports",
]
for f in FOLDERS:
    (ROOT / f).mkdir(parents=True, exist_ok=True)
    print("dibuat:", ROOT / f)
# .gitkeep agar folder kosong tetap terlacak Git
for f in FOLDERS:
    (ROOT / f / ".gitkeep").touch()
```

## Langkah 5. Memetakan posisi data acquisition dalam lifecycle
